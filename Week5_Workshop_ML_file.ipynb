## Step 6 · New columns with functions and conditions  *(2b)*
**Say:** "A function with if / elif / else turns a number into a group. `apply` runs it on every value; `apply(axis=1)` on every row."

**6.1** `bmi_group`

```python
def bmi_group(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25:
        return 'Normal'
    elif bmi < 30:
        return 'Overweight'
    else:
        return 'Obese'
```

**6.2** Apply it

```python
df.loc[:, 'BMI_group'] = df['BMI'].apply(bmi_group) #applies your bmi_group function to every value in the BMI column

df['BMI_group'].value_counts() #counts how many records belong to each BMI category
```

**6.3** `age_band`

```python
def age_band(age):
    if age < 30:
        return 'Under 30'
    elif age < 45:
        return '30-44'
    elif age < 60:
        return '45-59'
    else:
        return '60+'
```

**Extension:** lifestyle count (descriptive, not a prediction)

```python
def lifestyle_risk_count(row):

    count = 0

    if row['Physical_Activity'] == 'Sedentary':
        count += 1
    if row['Diet_Quality'] == 'Poor':
        count += 1
    if row['Alcohol_Consumption'] == 'Heavy':
        count += 1
    if row['Smoking_Status'] == 'Current':
        count += 1
    if row['Stress_Level'] in ['High', 'Very High']:
        count += 1
    return count
```

```python
df['Lifestyle_risk_count'] = df.apply(lifestyle_risk_count, axis=1)
df['Lifestyle_risk_count'].value_counts().sort_index()
```

### Lifestyle Risk Count

The `Lifestyle_risk_count` feature is created by checking five lifestyle-related risk factors for each person.

| Lifestyle Factor | Risk Condition | Score |
|---|---|---:|
| Physical Activity | Sedentary | 1 |
| Diet Quality | Poor | 1 |
| Alcohol Consumption | Heavy | 1 |
| Smoking Status | Current | 1 |
| Stress Level | High or Very High | 1 |

Each risk factor contributes **1 point if present** and **0 points if absent**.

The points are added together to create a total `Lifestyle_risk_count` between **0 and 5**.

For example, if a person is sedentary, has a poor diet, currently smokes, and has high stress, their lifestyle risk count will be **4**.

**Why do this?**

* Because instead of looking at five separate lifestyle columns every time, you create one summary variable that says:
“How many of these selected lifestyle risks does this person have?”

## Step 9 · Reusable code: functions  *(2b)*
**Say:** "The brief asks for readable, organised, reusable code. Wrap the steps in functions, then run them in a few lines."

**9.1** `load_data`

```python
def load_data(path):
    return pd.read_csv(path)
```

**9.2** `clean_data`

```python
def clean_data(data):
    data = data.drop_duplicates().copy()
    data['Gender'] = data['Gender'].replace({'F': 'Female', 'female': 'Female', 'M': 'Male', 'male': 'Male'})
    data['Waist_Circumference_cm'] = data['Waist_Circumference_cm'].fillna(data['Waist_Circumference_cm'].median())
    data['Sleep_Hours'] = data['Sleep_Hours'].fillna(data['Sleep_Hours'].median())
    data['Screen_Time_Hours'] = data['Screen_Time_Hours'].fillna(data['Screen_Time_Hours'].median())
    data['Cholesterol_mg_dL'] = data['Cholesterol_mg_dL'].fillna(data['Cholesterol_mg_dL'].median())
    data['Insulin_Level'] = data['Insulin_Level'].fillna(data['Insulin_Level'].median())
    data['Diet_Quality'] = data['Diet_Quality'].fillna(data['Diet_Quality'].mode()[0])
    data['Stress_Level'] = data['Stress_Level'].fillna(data['Stress_Level'].mode()[0])
    data['Alcohol_Consumption'] = data['Alcohol_Consumption'].fillna('None')
    return data
```

**9.3** `add_features`

```python
def add_features(data):
    data = data.copy()
    data['BMI_group'] = data['BMI'].apply(bmi_group)
    data['Age_band'] = data['Age'].apply(age_band)
    data['Risk_flag'] = (data['Diabetes_Risk'] == 'Yes').astype(int)
    return data
```

**9.4** Chain them. **Say:** "Three functions, one line, from raw file to analysis-ready table."

```python
df2 = add_features(clean_data(load_data('diabetes_risk_teaching_dataset_8000.csv')))
print(df2.shape, df2.isnull().sum().sum())
```
